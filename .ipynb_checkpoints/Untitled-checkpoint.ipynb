```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer

# 1. Liste de 3 phrases
phrases = [
    "Le joueur de football marque un but magnifique",
    "Ce film de cinéma est un chef d'œuvre magnifique",
    "Le joueur regarde un film de sport au cinéma",
]

# 2. Transformation avec CountVectorizer
vectorizer = CountVectorizer()
X_bow = vectorizer.fit_transform(phrases)

# 3. Récupération des mots du vocabulaire
mots = vectorizer.get_feature_names_out()

# Création d'un DataFrame Pandas pour un affichage clair
df_bow = pd.DataFrame(
    X_bow.toarray(),
    columns=mots,
    index=[f"Phrase {i+1}" for i in range(len(phrases))],
)

print("--- Mots extraits (Vocabulaire) ---")
print(mots)

print("\n--- Matrice Bag of Words ---")
print(df_bow)
```

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer

# 1. Utilisation de la même liste de phrases
phrases = [
    "Le joueur de football marque un but magnifique",
    "Ce film de cinéma est un chef d'œuvre magnifique",
    "Le joueur regarde un film de sport au cinéma",
]

# 2. Transformation avec TfidfVectorizer
tfidf_vectorizer = TfidfVectorizer()
X_tfidf = tfidf_vectorizer.fit_transform(phrases)

# 3. Création du DataFrame pour comparer les poids
mots_tfidf = tfidf_vectorizer.get_feature_names_out()
df_tfidf = pd.DataFrame(
    X_tfidf.toarray(),
    columns=mots_tfidf,
    index=[f"Phrase {i+1}" for i in range(len(phrases))],
)

print("--- Matrice TF-IDF (Poids arrondis) ---")
print(df_tfidf.round(3))
```
